```r
student_id <- readline("Enter Student ID: ")
attendance_str <- readline("Enter Attendance Grade: ")
grade_str <- readline("Enter Grade: ")

# Convert to numeric for calculations
attendance <- as.numeric(attendance_str)
grade <- as.numeric(grade_str)

classify_risk <- function(attendance_val, grade_val){
  # Handle non-numeric input for robustness
  if (is.na(attendance_val) || is.na(grade_val)) {
    return("Invalid Input")
  } else if (attendance_val <= 75 && grade_val <= 75){
    return("High Risk")
  } else if (attendance_val <= 85 || grade_val <= 80) {
    return("Moderate Risk")
  } else {
    return("Low Risk")
  }
}

recommend_action <- function(risk){
  if (risk == "High Risk"){return("Immediate academic intervention")
  } else if (risk == "Moderate Risk"){return("Academic monitoring")
  }else{return("Continue regular monitoring")
  }
}

cat("Student ID: ", student_id, "\n")
cat("Attendance: ", attendance, "%\n")
cat("Grade: ", grade, "\n")

risk_level <- classify_risk(attendance, grade)
cat("Risk Level: ", risk_level, "\n")

recommendation <- recommend_action(risk_level)
cat("Recommendation: ", recommendation, "\n")
```
