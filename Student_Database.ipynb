```r
students <- data.frame (
    Student_ID = c("S001", "S002", "S003", "S004", "S005", "S006", "S007", "S008"),
    Programming_Grade = c(88, 72, 91, 78, 65, 85, 74, 82),
    Math_Grade = c(85, 70, 88, 80, 68, 90, 73, 79),
    Attendance = c(92, 80, 95, 88, 70, 91, 76, 85),
    Prerequisite = c("Yes", "No", "Yes", "Yes", "No", "Yes", "No", "Yes")
)

write.csv(students, "Students.csv")

students
```

```r
prog <- student$Programming_Grade
cat("\n Programming Grade:", prog)
math <- student$Math_Grade
cat("\n Math Grade:", math)
att <- student$Attendance
cat("\n Attendance Grade:", att)
prereq <- student$Prerequisite
cat("\n Prerequisite:", prereq)
```

```r
evaluate_eligibility <- function(prog, math, att, prereq) {


if (prog >= 80 && math >= 75 && att >= 75) {
  return("Eligible")
} else if (prog >= 75 && math>= 70 && att >= 75) {
  return("Conditionally Eligible")
} else {
  return("Not Eligible")
}
}



```

```r
students$Eligibility <- mapply(
  evaluate_eligibility,
  students$Programming_Grade,
  students$Math_Grade,
  students$Attendance
)

students
```

```r
students_Recommendation <- function(evaluate_eligibility) {
  if(evaluate_eligibility == "Eligible"){
    return("Student May Enroll")
  } else if(evaluate_eligibility == "Conditional"){
    return("Consult academic tutor")
  } else{
    return("Complete requirements before enrollment.")
  }
}

students_Recommendation("Eligible")
```

```r
students$Recommendation <- mapply(
 students_Recommendation,
 students$Eligibility
)

students
```
