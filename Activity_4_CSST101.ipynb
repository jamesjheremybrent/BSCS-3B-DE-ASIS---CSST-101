```r
arguments <- data.frame (
  Argument_ID = c("A001", "A002", "A003", "A004", "A005", "A006"),
  Position = c("For", "For", "For", "Against", "Against", "Against"),
  Topic = c("Learning", "Productivity", "AI Literacy", "Academic Integrity", "Dependency", "Assessment"),
  Claim = c("AI can provide students with immediate learning support.", "AI can help students complete routine academic tasks more efficiently", "Allowing AI use can help students develop AI literacy.", "Undisclosed AI-generated work may create academic integrity concerns.", "Excessive AI use may reduce studets' independent problem-solving practice.", "AI use may make it more difficult to determine whether submitted work reflects student learning.")

)
```

```r
evidence <- data.frame (
  Evidence_ID = c("E001", "E002", "E003", "E004", "E005","E006"),
  Argument_ID = c("A001", "A002", "A003", "A004", "A005", "A006"),
  Evidence_Type = c("Research", "Research", "Policy", "Policy", "Research", "Research"),
  Strength = c("Strong", "Moderate", "Strong", "Strong", "Moderate", "Strong")
)
```

COUNT THE ARGUMENTS

```r
table(arguments$Position)
```

```r
table(evidence$Strength)
```

```r
nrow(arguments)
```

Retrieve Arguments Supporting the Policy

```r
for_arguments <- subset(
  arguments,
  Position == "For"
)

print(for_arguments)
```

RETRIEVE ARGUMENTS AGAINST THE POLICY

```r
for_arguments <- subset(
  arguments,
  Position == "Against"
)

print(for_arguments)
```

CONNECT ARGUMENTS AND EVIDENCE

```r
knowledge_base <- merge(
  arguments,
  evidence,
  by = "Argument_ID"
)

print(knowledge_base)
```

CONVERT EVIDENCE STRENGTH INTO A SCORE

```r
get_score <- function(Strength){

  if (Strength == "Strong"){
    return(3)
  }else if (Strength == "Moderate"){
    return(2)
  }else if (Strength == "Weak"){
    return(1)
  }else{
    return(0)
  }
}

get_score("Strong")
```

ADD EVIDENCE SCORES

```r
knowledge_base$Evidence_Score <- sapply(
  knowledge_base$Strength,
  get_score
)
print(knowledge_base)
```

CLASSIFY EVIDENCE SUPPORT

```r
classify_support <- function(score){
  if (score == 3){
    return("Strongly Supported")
  } else if (score == 2){
    return("Moderately Supported")
  } else if (score == 1){
    return("Weakly Supported")
  } else {
    return("No Support")
  }
}

classify_support(2)
```

SEARCH BY TOPIC

```r
search_topic <- function(topic){
  result <- subset(
    knowledge_base,
    Topic == topic
  )
    if(nrow(result)==0){
      cat("Topic not found.\n")
    } else {
      print(result)
    }
}
```

```r
search_topic("Learning")
```
